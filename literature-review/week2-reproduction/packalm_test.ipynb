```python
import sys
sys.path.insert(0, '/content/PackALM/pretrain')

from fairseq.models.roberta.model import RobertaEncoder
from fairseq.data import Dictionary
import torch

fake_dict = Dictionary()
for i in range(2178 - len(fake_dict)):
    fake_dict.add_symbol(f"tok{i}")
print("Sözlük boyutu:", len(fake_dict))

args = ckpt['args']
encoder = RobertaEncoder(args, fake_dict)
missing, unexpected = encoder.load_state_dict(ckpt['model'], strict=False)
print("Eksik anahtar sayısı:", len(missing))
print("Beklenmeyen anahtar sayısı:", len(unexpected))
```

```python
# checkpoint anahtarlarından 'decoder.' önekini kaldır
fixed_state_dict = {k.replace('decoder.', '', 1): v for k, v in ckpt['model'].items()}

missing, unexpected = encoder.load_state_dict(fixed_state_dict, strict=False)
print("Eksik anahtar sayısı:", len(missing))
print("Beklenmeyen anahtar sayısı:", len(unexpected))
if missing:
    print("İlk birkaç eksik:", missing[:5])
if unexpected:
    print("İlk birkaç beklenmeyen:", unexpected[:5])
```

```python
import torch

encoder.eval()
# Sentetik girdi: batch=1, sequence_length=20, sözlük aralığında rastgele token ID'leri
fake_input = torch.randint(low=4, high=2178, size=(1, 20))  # 0-3 genelde özel token'lar (pad, bos, eos, unk)

with torch.no_grad():
    output = encoder(fake_input, features_only=True)

if isinstance(output, tuple):
    features = output[0]
else:
    features = output

print("Çıktı şekli:", features.shape)
print("Örnek değerler (ilk token, ilk 5 boyut):", features[0, 0, :5])
```
